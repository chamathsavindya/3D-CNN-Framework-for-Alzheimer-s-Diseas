```python
import os, glob, shutil, subprocess

IN_ROOT  = "/content/drive/MyDrive/hippodeep_in"
OUT_ROOT = "/content/drive/MyDrive/hippodeep_out"

def run_all(label):
    in_dir = os.path.join(IN_ROOT, label)
    out_dir = os.path.join(OUT_ROOT, label)
    os.makedirs(out_dir, exist_ok=True)

    files = sorted(glob.glob(os.path.join(in_dir, "*.nii.gz")))
    print(label, ":", len(files), "files")

    for f in files:
        subprocess.run(["python", "hippodeep.py", f], check=True)

        base = f[:-7]  # remove ".nii.gz"
        outputs = [
            base + "_mask_L.nii.gz",
            base + "_mask_R.nii.gz",
            base + "_brain_mask.nii.gz",
            base + "_hippoLR_volumes.csv",
        ]

        for p in outputs:
            if os.path.exists(p):
                shutil.move(p, os.path.join(out_dir, os.path.basename(p)))

run_all("AD")
run_all("CN")
print("All done ✅")
```
